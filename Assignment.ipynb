Q1. Read an image using OpenCV and display it


Approach:

Use cv2.imread() to read the image and cv2.imshow()/Matplotlib to display it.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('CAT_images.jpg')

plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

Observation:
The image was successfully read using OpenCV and displayed using Matplotlib after converting the color format from BGR to RGB.

Result:
The input image was successfully loaded and visualized.

Q14. Split a color image into its B, G, and R channels and display each channel.

Approach:
Use cv2.split() and display each single-channel image with a grayscale colormap.

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread('CAT_images.jpg')

b, g, r = cv2.split(image)

blue = np.zeros_like(image)
green = np.zeros_like(image)
red = np.zeros_like(image)

blue[:, :, 0] = b
green[:, :, 1] = g
red[:, :, 2] = r

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(blue, cv2.COLOR_BGR2RGB))
plt.title("Blue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(green, cv2.COLOR_BGR2RGB))
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(red, cv2.COLOR_BGR2RGB))
plt.title("Red Channel")
plt.axis("off")

plt.show()
```

Observation
Each channel shows the intensity contribution of one color component.

Result
The B, G, and R channels were successfully separated and displayed.

Q15. Merge separate B, G, and R channels into a single color image.

Approach:
Split the source image, merge the channels with cv2.merge(), and convert BGR to RGB for display.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('CAT_images.jpg')
b, g, r = cv2.split(image)
merged = cv2.merge([b, g, r])
plt.imshow(cv2.cvtColor(merged, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

Result
The channels were successfully merged into a single color image.

Q21. Convert an 8-bit grayscale image into a 4-bit quantized image.

Approach:
Reduce the 256 levels to 16 levels using integer division by 16.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('CAT_images.jpg', cv2.IMREAD_GRAYSCALE)
quantized = (image // 16) * 16
plt.imshow(quantized, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.show()
```

Observation
The image contains 16 discrete intensity levels instead of 256.

Result
The 8-bit image was successfully converted to a 4-bit quantized image.

Q25. Rotate an image by 90 degrees, display it, and save it.

Approach:
Use cv2.rotate() for a clockwise 90-degree rotation, then display and save the result.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('CAT_images.jpg')
rotated = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
cv2.imwrite('rotated_image.jpg', rotated)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```
